
# **K-Nearest Neighbours (KNN)**

El modelo de K-Nearest Neighbours (KNN) es un algoritmo de aprendizaje automático no paramétrico y basado en instancias. Se utiliza para clasificación y regresión. En ambos casos, la entrada consiste en los k vecinos más cercanos de la muestra a clasificar o predecir.



## Invocación del Modelo KNN en Scikit-Learn

Para utilizar el modelo KNN en Scikit-Learn, primero debemos importar la clase `KNeighborsClassifier` para clasificación o `KNeighborsRegressor` para regresión desde el módulo `neighbors`. Luego, creamos una instancia del modelo y la entrenamos con los datos.

```python
from sklearn.neighbors import KNeighborsClassifier

# Para clasificación
knn_clas = KNeighborsClassifier(n_neighbors=5)
knn_clas.fit(X_train, y_train)

# Para regresión
from sklearn.neighbors import KNeighborsRegressor
knn_reg = KNeighborsRegressor(n_neighbors=5)
knn_reg.fit(X_train, y_train)
```



## Hiperparámetros Importantes en KNN

Algunos de los hiperparámetros más destacados en el modelo KNN son:

- `n_neighbors`: Número de vecinos a considerar. Es el parámetro principal que define el modelo.
- `weights`: Función de peso utilizada en la predicción. Puede ser 'uniform' (todos los vecinos tienen el mismo peso) o 'distance' (los vecinos más cercanos tienen mayor influencia).
- `metric`: La métrica de distancia utilizada para el árbol. Por defecto, se utiliza la distancia euclidiana ('minkowski' con p=2).
- `algorithm`: Algoritmo utilizado para calcular los vecinos más cercanos. Puede ser 'auto', 'ball_tree', 'kd_tree', o 'brute'.



## Hiperparámetros para Optimización con GridSearch o RandomSearch

Los hiperparámetros que comúnmente se optimizan en KNN a través de técnicas como GridSearch o RandomSearch incluyen:

- `n_neighbors`: Generalmente se prueba con un rango de valores para encontrar el número óptimo de vecinos.
- `weights`: Se puede experimentar con 'uniform' y 'distance' para ver cuál ofrece mejores resultados.
- `metric`: Aunque la distancia euclidiana es común, se pueden probar otras métricas como la distancia de Manhattan ('manhattan') o Chebyshev ('chebyshev').



## Sugerencias para Aplicar el Modelo KNN

El modelo KNN es particularmente útil en situaciones donde:

- Los datos tienen pocas dimensiones (espacio de baja dimensionalidad), ya que KNN puede sufrir con la "maldición de la dimensionalidad".
- La relación entre los atributos es intuitivamente cercana, lo que hace que la distancia sea una buena medida de similitud.
- Se prefiere un modelo simple sin la necesidad de ajustar muchos parámetros.

Sin embargo, se debe tener cuidado con grandes conjuntos de datos, ya que KNN puede ser computacionalmente costoso.
