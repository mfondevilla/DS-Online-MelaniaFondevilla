```python
from matplotlib.ticker import FuncFormatter
# plt.rcParams['text.usetex'] = True

fig = plt.figure(figsize=(20,20))

plt.subplot(2,2,1)
top5population = df.nlargest(5, "population_total")["population_total"]
top5cities = df.nlargest(5, "population_total")["city"]
colors = plt.cm.Pastel1(np.linspace(0, 1, 5))
plt.bar(top5cities, top5population, color = ["blue", "grey", "red","#808000","brown"], alpha = 0.5)
plt.ylabel("Población (miles de habitantes)")
# Etiquetas encima de cada barra
for i, v in enumerate(top5population):
    plt.text(i, v, f"{v/1000:.2f}", ha='center', va='bottom', fontsize=10)
plt.ticklabel_format(style='plain', axis='y') # desactivar notación científica
def miles(x, pos):
    return f"{int(x/1000)}"
plt.gca().yaxis.set_major_formatter(FuncFormatter(miles))
plt.title("Ciudades más pobladas de California")



plt.subplot(2,2,2)
plt.hist(df.area_land_km2, bins=50, color="steelblue", alpha = 0.7, label="Hist Area")
# añadir fontdict
plt.legend(loc = "upper right", prop={"weight": "bold"})
plt.ylabel("Número de Ciudades", family= "Arial", weight = "bold")
plt.xlabel(r"$\mathbf{Superficie}\;(\mathit{km}^{2})$")
plt.title("Histograma del área total ocupada por las ciudades")

plt.subplot(2,2,3)
population = df.population_total
km_agua = df.area_water_km2
plt.scatter(km_agua, population, color= "green", alpha=0.5);
plt.ylabel("Población (miles habitantes)")
plt.xlabel(r"Área del agua en la ciudad$(\mathit{km}^{2})$")
plt.ticklabel_format(style='plain', axis='y')
def miles(x, pos):
    return f"{int(x/1000)}"
plt.gca().yaxis.set_major_formatter(FuncFormatter(miles))

#  outliers
outlier_1_y = df['population_total'].nlargest(2).iloc[0]
outlier_2_y = df['population_total'].nlargest(2).iloc[1]
outlier_1_x = df['area_water_km2'].nlargest(2).iloc[0]
outlier_2_x = df['area_water_km2'].nlargest(2).iloc[1]
# repasar los outliers porque sinecesitamos valores únicops
''' 
origenx = pd.Series(y, index= x).idxmin()
origeny = pd.Series(y, index= x).min()
to_x = population.max()
to_y = km_agua.max()
props_flecha = {
    "arrowstyle": "<->",
    "connectionstyle": "angle3,angleA=0,angleB=90"
}
plt.annotate("Outliers", xy=(origenx, origeny), xytext=(to_x, to_y))
'''
plt.title("Población por km cuadrado de agua")

plt.subplot(2,2,4)
elevations = df.elevation_m.dropna()
plt.boxplot(elevations, 
            vert = False,
            patch_artist=True,
            boxprops = {'facecolor': 'red', 'edgecolor': 'black'},
            medianprops = {'color': 'black'})
plt.yticks([]) # elimina ticks
plt.title("Diagrama de caja de la altitud de las Ciudades de California")


fig.subplots_adjust()
```
